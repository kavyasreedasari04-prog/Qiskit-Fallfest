```python
print("CHALLENGE 4: PREDICTIONS BEFORE EXECUTION")

print("\nPart A - Correlated Outcomes")
print("Expected: 00 and 11, approximately 50% each.")

print("\nPart B - Opposite Outcomes")
print("Expected: 01 and 10, approximately 50% each.")

print("\nReason:")
print("An X gate on one qubit flips its measurement value.")
```

```python
from qiskit import QuantumCircuit
from IPython.display import display

# Part A: Correlated outcomes
qc_a = QuantumCircuit(2, 2)

qc_a.h(0)
qc_a.cx(0, 1)

# Measure both qubits
qc_a.measure([0, 1], [0, 1])

print("Part A: Correlated Outcomes")
display(qc_a.draw("mpl"))
```

```python
# Part B: Opposite outcomes
qc_b = QuantumCircuit(2, 2)

qc_b.h(0)
qc_b.cx(0, 1)

# Add one X gate to flip qubit 0
qc_b.x(0)

# Measure both qubits
qc_b.measure([0, 1], [0, 1])

print("Part B: Opposite Outcomes")
display(qc_b.draw("mpl"))
```

```python
from qiskit import transpile
from qiskit_aer import AerSimulator

# Create simulator
simulator = AerSimulator()

# Run Part A for 100 shots
result_a = simulator.run(
    transpile(qc_a, simulator),
    shots=100
).result()

# Run Part B for 100 shots
result_b = simulator.run(
    transpile(qc_b, simulator),
    shots=100
).result()

# Get measurement results
counts_a = result_a.get_counts()
counts_b = result_b.get_counts()

print("Part A - Correlated Outcomes:")
print(counts_a)

print("\nPart B - Opposite Outcomes:")
print(counts_b)
```

```python
from qiskit.visualization import plot_histogram
from IPython.display import display

display(
    plot_histogram(
        [counts_a, counts_b],
        legend=["Part A: Correlated", "Part B: Opposite"],
        title="Challenge 4: Correlated vs Opposite Outcomes"
    )
)
```

```python
print("CHALLENGE 4: COMPARISON AND EXPLANATION")

print("\nPart A - Correlated outcomes:")
print(counts_a)

print("\nPart B - Opposite outcomes:")
print(counts_b)

print("""
Explanation:
Part A uses an H gate followed by a CX gate to create a Bell state.
The measurement outcomes are correlated, so 00 and 11 dominate.

Part B adds one X gate to qubit 0.
The X gate flips that qubit, changing the dominant outcomes to 01 and 10.

Therefore, adding one quantum gate changes the relationship between
the measurement outcomes from matching to opposite.
""")
```
