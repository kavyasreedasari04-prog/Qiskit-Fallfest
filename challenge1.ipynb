```python
from qiskit import QuantumCircuit, transpile
from qiskit_aer import AerSimulator
from qiskit.visualization import plot_histogram
import matplotlib.pyplot as plt

simulator = AerSimulator()

print("Qiskit libraries imported successfully!")
```

```python
# Circuit A: One X gate
qc_a = QuantumCircuit(1, 1)
qc_a.x(0)
qc_a.measure(0, 0)

# Circuit B: Two X gates
qc_b = QuantumCircuit(1, 1)
qc_b.x(0)
qc_b.x(0)
qc_b.measure(0, 0)

# Circuit C: One H gate
qc_c = QuantumCircuit(1, 1)
qc_c.h(0)
qc_c.measure(0, 0)

print("Circuit A: One X gate")
display(qc_a.draw("mpl"))

print("Circuit B: Two X gates")
display(qc_b.draw("mpl"))

print("Circuit C: One H gate")
display(qc_c.draw("mpl"))
```

```python
shots = 100

# Run Circuit A
result_a = simulator.run(
    transpile(qc_a, simulator), shots=shots
).result()

# Run Circuit B
result_b = simulator.run(
    transpile(qc_b, simulator), shots=shots
).result()

# Run Circuit C
result_c = simulator.run(
    transpile(qc_c, simulator), shots=shots
).result()

# Get measurement counts
counts_a = result_a.get_counts()
counts_b = result_b.get_counts()
counts_c = result_c.get_counts()

print("Circuit A results:", counts_a)
print("Circuit B results:", counts_b)
print("Circuit C results:", counts_c)
```

```python
display(plot_histogram(counts_a, title="Circuit A: One X Gate"))

display(plot_histogram(counts_b, title="Circuit B: Two X Gates"))

display(plot_histogram(counts_c, title="Circuit C: One H Gate"))
```

```python
print("COMPARISON OF PREDICTIONS AND RESULTS")
print("-------------------------------------")

print("\nCircuit A: One X gate")
print("Prediction: 100 measurements of 1")
print("Actual result:", counts_a)

print("\nCircuit B: Two X gates")
print("Prediction: 100 measurements of 0")
print("Actual result:", counts_b)

print("\nCircuit C: One H gate")
print("Prediction: Approximately 50 measurements of 0 and 50 of 1")
print("Actual result:", counts_c)
```
