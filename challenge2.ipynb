```python
import math

print("PREDICTIONS BEFORE EXECUTION")
print("Circuit: One RY gate with angle pi/3")
print("Prediction: Outcome 0 will occur more frequently.")
print("Predicted probability of 0: 75%")
print("Predicted probability of 1: 25%")
```

```python
from qiskit import QuantumCircuit
from IPython.display import display
import math

# Create a circuit with exactly 1 qubit and 1 classical bit
qc = QuantumCircuit(1, 1)

# Apply RY gate with angle pi/3
qc.ry(math.pi / 3, 0)

# Measure the qubit
qc.measure(0, 0)

# Display the circuit
display(qc.draw("mpl"))
```

```python
from qiskit import transpile
from qiskit_aer import AerSimulator

# Create the simulator
simulator = AerSimulator()

# Execute the circuit for 100 shots
compiled_circuit = transpile(qc, simulator)
result = simulator.run(compiled_circuit, shots=100).result()

# Get measurement counts
counts = result.get_counts()

print("Measurement results for 100 shots:")
print(counts)
```

```python
from qiskit.visualization import plot_histogram
from IPython.display import display

display(
    plot_histogram(
        counts,
        title="Controlled Measurement Probabilities - 100 Shots"
    )
)
```

```python
print("PREDICTED PROBABILITIES")
print("Outcome 0: 75%")
print("Outcome 1: 25%")

print("\nACTUAL RESULTS")
print(counts)

print("\nACTUAL PERCENTAGES")
for outcome in ["0", "1"]:
    count = counts.get(outcome, 0)
    print(f"Outcome {outcome}: {count}%")
```

```python
from qiskit import QuantumCircuit, transpile
from qiskit_aer import AerSimulator
from qiskit.visualization import plot_histogram
from IPython.display import display

simulator = AerSimulator()

# Bonus Circuit A: Outcome 0 occurs more frequently
qc_bonus_a = QuantumCircuit(1, 1)
qc_bonus_a.ry(3.14159 / 3, 0)
qc_bonus_a.measure(0, 0)

# Bonus Circuit B: Outcome 1 occurs more frequently
qc_bonus_b = QuantumCircuit(1, 1)
qc_bonus_b.ry(2 * 3.14159 / 3, 0)
qc_bonus_b.measure(0, 0)

print("Bonus Circuit A")
display(qc_bonus_a.draw("mpl"))

print("Bonus Circuit B")
display(qc_bonus_b.draw("mpl"))
```

```python
result_bonus_a = simulator.run(
    transpile(qc_bonus_a, simulator), shots=100
).result()

result_bonus_b = simulator.run(
    transpile(qc_bonus_b, simulator), shots=100
).result()

counts_bonus_a = result_bonus_a.get_counts()
counts_bonus_b = result_bonus_b.get_counts()

print("Bonus Circuit A:", counts_bonus_a)
print("Bonus Circuit B:", counts_bonus_b)
```

```python
display(plot_histogram(
    [counts_bonus_a, counts_bonus_b],
    legend=["Circuit A: More zeros", "Circuit B: More ones"],
    title="Bonus Challenge: Comparing Measurement Bias"
))
```
