```python
print("CHALLENGE 3: PREDICTION BEFORE EXECUTION")
print()
print("Original Bell circuit:")
print("Expected outcomes: 00 and 11, approximately 50% each.")
print()
print("Modified Bell circuit:")
print("Prediction: Applying an X gate to one qubit")
print("will change the correlation.")
print("Expected outcomes: 01 and 10, approximately 50% each.")
```

```python
from qiskit import QuantumCircuit
from IPython.display import display

# Original Bell-state circuit
qc_original = QuantumCircuit(2, 2)

qc_original.h(0)
qc_original.cx(0, 1)

# Measure both qubits
qc_original.measure([0, 1], [0, 1])

display(qc_original.draw("mpl"))
```

```python
# Modified Bell-state circuit
qc_modified = QuantumCircuit(2, 2)

qc_modified.h(0)
qc_modified.cx(0, 1)

# Add one X gate to qubit 0
qc_modified.x(0)

# Measure both qubits
qc_modified.measure([0, 1], [0, 1])

print("Modified Bell-state circuit:")
display(qc_modified.draw("mpl"))
```

```python
from qiskit import transpile
from qiskit_aer import AerSimulator

# Create simulator
simulator = AerSimulator()

# Run original Bell circuit for 100 shots
result_original = simulator.run(
    transpile(qc_original, simulator),
    shots=100
).result()

# Run modified Bell circuit for 100 shots
result_modified = simulator.run(
    transpile(qc_modified, simulator),
    shots=100
).result()

# Get measurement counts
counts_original = result_original.get_counts()
counts_modified = result_modified.get_counts()

print("Original Bell circuit results:")
print(counts_original)

print("\nModified Bell circuit results:")
print(counts_modified)
```

```python
from qiskit.visualization import plot_histogram
from IPython.display import display

display(
    plot_histogram(
        [counts_original, counts_modified],
        legend=["Original Bell Circuit", "Modified Bell Circuit"],
        title="Challenge 3: Measurement Correlation"
    )
)
```

```python
print("CHALLENGE 3: RESULTS AND EXPLANATION")
print("\nOriginal Bell circuit:")
print(counts_original)

print("\nModified Bell circuit:")
print(counts_modified)

print("""
Explanation:
The original Bell circuit uses an H gate followed by a CX gate.
It creates correlated measurement outcomes, so 00 and 11 occur.

The modified circuit adds one X gate to qubit 0.
The X gate flips that qubit, changing the outcomes from 00 and 11
to 01 and 10. The qubits remain correlated, but their measurement
values are now opposite.
""")
```
